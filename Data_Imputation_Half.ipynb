### Import Packages

### Read in the Dataset

```python
# Point to data files location
path = "./scraper/data/6_23_2025/"
all_files = glob.glob(path + "*.json")

# Read all files into single dataframe
df_list = []
for file in all_files:
    df = pd.read_json(file)
    df_list.append(df)

# ignore_index resets index of combined DF to prevent index duplicates
raw_df = pd.concat(df_list, ignore_index=True) 

# Display entire data frame
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)

# Examine and excerpt of data
raw_df.head(5)
```

### Drop Duplicate Rows

```python
# Make a copy of the original dataframe
df = raw_df.copy()

# Check rows with duplicate VINS and drop
df = df.drop_duplicates(subset=['vin'], keep='first')
```

### Convert Engine Size to Numeric Datatype

```python
# Convert Engine Size into a numeric feature by removing the L and changing datatype
df['engineSize'] = df['engineSize'].str.replace('L', '', regex=False)
df['engineSize'] = pd.to_numeric(df['engineSize'])

print(df['engineSize'].dtype)
```

```python
# Remove all outliers in the mileage column based on a zscore of 3
df = df[np.abs(stats.zscore(df['mileage'])) < 3]

print("DataFrame Shape After:", df.shape)
```

```python
# Use method to display a list of missing values and their percentages

# credit: https://www.kaggle.com/willkoehrsen/start-here-a-gentle-introduction. 

def missing_values_table(df):
    # Total missing values
    mis_val = df.isnull().sum()
    
    # Percentage of missing values
    mis_val_percent = 100 * df.isnull().sum() / len(df)
    
    # Make a table with the results
    mis_val_table = pd.concat([mis_val, mis_val_percent], axis=1)
    
    # Rename the columns
    mis_val_table_ren_columns = mis_val_table.rename(
    columns = {0 : 'Missing Values', 1 : '% of Total Values'})
    
    # Sort the table by percentage of missing descending
    mis_val_table_ren_columns = mis_val_table_ren_columns[
        mis_val_table_ren_columns.iloc[:,1] != 0].sort_values(
    '% of Total Values', ascending=False).round(2)
    
    # Print some summary information
    print ("Your selected dataframe has " + str(df.shape[1]) + " columns.\n"      
        "There are " + str(mis_val_table_ren_columns.shape[0]) +
            " columns that have missing values.")
    
    # Return the dataframe with missing information
    return mis_val_table_ren_columns
```

```python
# Apply missing values table method on the dataframe
missing_vals = missing_values_table(df)
missing_vals
```

### Impute Feature Scores and Trims

```python
# Regressor for Numerical Values
# Classifier for Categorical Values
from lightgbm import LGBMRegressor, LGBMClassifier
```

```python
# Display columns in df with missing vals
cols_with_nans = [x for x in df if df[x].isnull().sum() > 0]
cols_with_nans
```

```python
# Divide into categorical and numeric columns
num_cols = [col for col in df[cols_with_nans].select_dtypes(exclude='O')]
cat_cols = [col for col in df[cols_with_nans].select_dtypes(include='O')]

print("Numeric Columns: \n", num_cols)
print("Categorical Columns: \n", cat_cols)
```

```python

def factorize(data):
    # Factorize categorical data so its represented as integers
    for cat_col in data.select_dtypes(include='O'):
        data[cat_col] = pd.factorize(data[cat_col])[0]
    return data

def create_splits(data, col):
    # Train Test split on null values present
    train = data[data['is_nan'] == 0]
    test = data[data['is_nan'] == 1]
    X_train = train.drop([col, 'is_nan'], axis=1)
    y_train = train[col]
    X_test = test.drop([col, 'is_nan'], axis=1)
    return X_train, y_train, X_test

def train_predict(mode, X_train, y_train, X_test):
    if mode == 'regression':
        model = LGBMRegressor()
    else :
        model = LGBMClassifier()
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    return y_pred


def impute_missing(df, col_list, mode):
    for col in col_list:
        print("Column Examined: ", col)
        if df[col].isnull().sum() > int(len(df)/2):
            df.drop(col, axis=1, inplace=True)
            print(f'Dropped {col} because it had over 50% missing values')
        else:
            data = df.copy()
            # Reset Indexes
            data = data.reset_index(drop=True)
            nan_ixs = np.where(data[col].isna())[0]
            data['is_nan'] = 0
            data.loc[nan_ixs, 'is_nan'] = 1
            X = data.drop([col], axis=1)
            y = data[col]
            X = factorize(X)
            data = X.join(y)
            X_train, y_train, X_test = create_splits(data, col)
            y_pred = train_predict(mode, X_train, y_train, X_test)
            # If prediction is numeric then round it
            if y_pred.dtype == 'float64':
                y_pred = np.round(y_pred)
            print("Predictions: \n", y_pred)
            # Include predicted values in the indices of missing column
            df = df.reset_index(drop=True)
            df.loc[nan_ixs, col] = y_pred
            # df = data
    return df

df = impute_missing(df, num_cols, 'regression')
df = impute_missing(df, cat_cols, 'classification')
        
```

### Calculate Variance Inflation Factor (VIF)

```python
# Import VIF packages
import pandas as pd
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant
```

```python
# Create numeric df
numeric_df_half = df.select_dtypes(include='number')

# Add constant in order to run VIF
num_df_con_half = add_constant(numeric_df_half)

# Remove missing values
num_df_con_half = num_df_con_half.copy().dropna()
```

```python
# Create a DataFrame to store VIF values
vif_data_half = pd.DataFrame()
vif_data_half['Feature'] = num_df_con_half.columns
vif_data_half['VIF'] = [variance_inflation_factor(num_df_con_half.values, i) for i in range(num_df_con_half.shape[1])]
vif_data_half = vif_data_half.round(2)
print("DataFrame Shape:", num_df_con_half.shape)
print(vif_data_half)
```

### Multicollinearity

Multicollinearity occurs when highly correlated variables contain similar information about the variance in the target variable, which in this case is basePrice. If multicollinearity issues are not addressed, the regression models produced will be susceptible to interpretability issues (the individual effects of each predictor will be hard to decipher) and small changes in the data could lead to large changes in the estimated coefficients.

### Why PCA?

Principal Component Analysis is an unsupervised pre-processing task that can help address multicollinearity issues by merging highly correlated variables into a set of uncorrelated variables.

### Disadvantage

Unable to explain the model's decisions with the uncorrelated feature set produced by PCA

### Questions
(1) Should PCA be performed on a subset of the highlight correlated features or on the entire dataset?

(2) Is the entire dataset different after pca or do certain features remain the same?

Answer: In general, after Principal Component Analysis (PCA) is applied, the entire dataset is transformed into a new set of values, and no original features remain exactly the same.
